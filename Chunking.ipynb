```python
from pypdf import PdfReader

doc=PdfReader("G:/My Drive/Yuvi Obsidian/2. Areas/Semester 5/__Support/Cloud Computing_1st.pdf")
page=doc.pages[0]
text=page.extract_text(extraction_mode="layout")
len(page.get_contents().get_data())
```

```python
print(text)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=0)
chunks=text_splitter.split_text(text)
```

```python
count=1
for chunk in chunks:
    print(f"{count}. {chunk}")
    count+=1
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter2=CharacterTextSplitter.from_tiktoken_encoder(encoding_name="cl100k_base", chunk_size=100,chunk_overlap=0)

chunks2=text_splitter2.split_text(text)
```

```python
count=1
for chunk in chunks2:
    print(f"{count}. {chunk}")
    count+=1
```

```python
text_splitter3=RecursiveCharacterTextSplitter.from_tiktoken_encoder(model_name="gpt-4",chunk_size=100,chunk_overlap=10)
chunks3=text_splitter3.split_text(text)
```

```python
count=1
for chunk in chunks3:
    print(f"{count}. {chunk}")
    count+=1
```
